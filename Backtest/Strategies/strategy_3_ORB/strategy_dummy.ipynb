```python
import pandas as pd

orb_range_time = "10:00"
DATA_PATH = "D:/FM/ALGOTRADING/artifacts/backtest/hist_data_5.json"

data = pd.read_json(DATA_PATH)

# Normalize date/time
data["date"] = pd.to_datetime(data["date"]).dt.date
data["time"] = pd.to_datetime(data["time"].astype(str)).dt.time

orb_start = pd.to_datetime("09:15").time()
orb_end = pd.to_datetime(orb_range_time).time()

# ORB window
orb = data[
    (data["time"] >= orb_start) &
    (data["time"] < orb_end)
]

# Calculate ORB separately for EACH symbol on EACH day
orb_daily = (
    orb.groupby(["date", "symbol"])
       .agg(
           orb_high=("high", "max"),
           orb_low=("low", "min")
       )
       .reset_index()
)

orb_daily["orb_range"] = (
    orb_daily["orb_high"] - orb_daily["orb_low"]
)

# Merge back using BOTH date and symbol
data = data.merge(
    orb_daily,
    on=["date", "symbol"],
    how="left"
)

data
```
